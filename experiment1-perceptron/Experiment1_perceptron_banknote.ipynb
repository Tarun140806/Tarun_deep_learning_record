```python
#TASK 3
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

feature_cols = ["variance", "skewness", "curtosis", "entropy"]

X = df[feature_cols].values
y = df["class"].values

# Normalize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])
```

```python
#TASK 4 & 5
import numpy as np

n_features = X_train.shape[1]
rng = np.random.default_rng(42)
w = rng.normal(0, 0.01, size=n_features)
b = 0.0

class SingleLayerPerceptron:
    def __init__(self, n_features, learning_rate=0.01, epochs=50, random_state=42):
        rng = np.random.default_rng(random_state)
        self.w = rng.normal(0, 0.01, size=n_features)
        self.b = 0.0
        self.lr = learning_rate
        self.epochs = epochs

    @staticmethod
    def step_activation(z):
        return np.where(z >= 0, 1, 0)

    def forward(self, x_row):
        z = np.dot(self.w, x_row) + self.b
        return self.step_activation(z)

    def predict(self, X):
      z = X @ self.w + self.b   # matrix-vector product: does the dot product for every row at once
      return self.step_activation(z)

    def fit(self, X, y, verbose=True):
      n_samples = X.shape[0]
      self.history = []  # to log epoch-wise errors/weights/bias

      for epoch in range(1, self.epochs + 1):
          misclassified = 0
          for i in range(n_samples):
              x_i = X[i]
              y_i = y[i]
              y_hat = self.forward(x_i)
              error = y_i - y_hat

              if error != 0:
                  misclassified += 1
                  self.w += self.lr * error * x_i
                  self.b += self.lr * error

          self.history.append({
              "epoch": epoch,
              "errors": misclassified,
              "w": self.w.copy(),
              "b": self.b
          })

          if verbose:
              print(f"Epoch {epoch:3d} | Misclassified: {misclassified:4d} | "
                    f"w: {self.w} | b: {self.b:.4f}")

          if misclassified == 0:
              print(f"Converged at epoch {epoch}")
              break
```

```python
model = SingleLayerPerceptron(n_features=X_train.shape[1])
prediction = model.forward(X_train[0])
print("Predicted:", prediction, "| Actual:", y_train[0])
```

```python
#TASK 6
y_pred = model.predict(X_test)
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print(f"Accuracy : {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1-score : {f1:.4f}")
print("Confusion Matrix:\n", cm)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_decision_boundary(w, b, X, y, title, ax):
    ax.scatter(X[:,0], X[:,1], c=y, cmap="bwr", s=150, edgecolors="k", zorder=3)

    x_vals = np.array([-0.5, 1.5])
    if w[1] != 0:
        y_vals = -(w[0]*x_vals + b) / w[1]
        ax.plot(x_vals, y_vals, "g-", linewidth=2)
    elif w[0] != 0:
        ax.axvline(x=-b/w[0], color="g", linewidth=2)

    ax.set_xlim(-0.5, 1.5)
    ax.set_ylim(-0.5, 1.5)
    ax.set_xlabel("x1")
    ax.set_ylabel("x2")
    ax.set_title(title, fontsize=9)
    ax.grid(alpha=0.3)
```

```python
class Perceptron:
    def __init__(self, n_features, learning_rate=0.1):
        self.w = np.zeros(n_features)
        self.b = 0.0
        self.lr = learning_rate
        self.history = []

    def step(self, z):
        return 1 if z >= 0 else 0

    def predict(self, x):
        z = np.dot(self.w, x) + self.b
        return self.step(z)

    def fit(self, X, y, epochs=10):
        self.history.append((self.w.copy(), self.b, "initial"))
        for epoch in range(epochs):
            total_errors = 0
            for i in range(len(X)):
                y_hat = self.predict(X[i])
                error = y[i] - y_hat
                if error != 0:
                    self.w += self.lr * error * X[i]
                    self.b += self.lr * error
                    total_errors += 1
                    self.history.append((self.w.copy(), self.b,
                                          f"epoch {epoch+1}, sample {i}"))
            if total_errors == 0:
                print(f"Converged after epoch {epoch+1}")
                break
```

```python
def train_and_plot(X, y, gate_name, learning_rate=0.1, epochs=10):
    p = Perceptron(n_features=X.shape[1], learning_rate=learning_rate)
    p.fit(X, y, epochs=epochs)

    n_updates = len(p.history)
    n_cols = 4
    n_rows = int(np.ceil(n_updates / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(4*n_cols, 4*n_rows))
    axes = axes.flatten() if n_updates > 1 else [axes]

    for idx, (w, b, tag) in enumerate(p.history):
        plot_decision_boundary(w, b, X, y, f"{tag}\nw={np.round(w,2)}, b={b:.2f}", axes[idx])

    for idx in range(n_updates, len(axes)):
        axes[idx].axis("off")

    fig.suptitle(f"{gate_name} Gate — Decision Boundary After Each Update", fontsize=14)
    plt.tight_layout()
    plt.show()

    return p

# AND gate
X_and = np.array([[0,0],[0,1],[1,0],[1,1]])
y_and = np.array([0,0,0,1])
p_and = train_and_plot(X_and, y_and, "AND")

# OR gate
X_or = np.array([[0,0],[0,1],[1,0],[1,1]])
y_or = np.array([0,1,1,1])
p_or = train_and_plot(X_or, y_or, "OR")
```

```python
def plot_decision_boundary_1d(w, b, X, y, title, ax):
    ax.scatter(X[:,0], np.zeros_like(X[:,0]), c=y, cmap="bwr", s=150, edgecolors="k", zorder=3)
    if w[0] != 0:
        boundary = -b / w[0]
        ax.axvline(x=boundary, color="g", linewidth=2)
    ax.set_xlim(-0.5, 1.5)
    ax.set_ylim(-1, 1)
    ax.set_xlabel("x1")
    ax.set_yticks([])
    ax.set_title(title, fontsize=9)
    ax.grid(alpha=0.3)

def train_and_plot_1d(X, y, gate_name, learning_rate=0.1, epochs=10):
    p = Perceptron(n_features=X.shape[1], learning_rate=learning_rate)
    p.fit(X, y, epochs=epochs)

    n_updates = len(p.history)
    fig, axes = plt.subplots(1, n_updates, figsize=(4*n_updates, 3))
    axes = axes if n_updates > 1 else [axes]

    for idx, (w, b, tag) in enumerate(p.history):
        plot_decision_boundary_1d(w, b, X, y, f"{tag}\nw={np.round(w,2)}, b={b:.2f}", axes[idx])

    fig.suptitle(f"{gate_name} Gate — Decision Boundary After Each Update", fontsize=14)
    plt.tight_layout()
    plt.show()

    return p

X_not = np.array([[0],[1]])
y_not = np.array([1,0])
p_not = train_and_plot_1d(X_not, y_not, "NOT")
```
